```python
import os, glob, pickle, sys, types
import numpy as np
import torch

BUFFER_DIR = os.path.join(os.path.dirname(os.path.abspath('.')), "acmi_records")
if not os.path.isdir(BUFFER_DIR):
    BUFFER_DIR = os.path.join(os.getcwd(), "acmi_records")
print('Buffer dir:', BUFFER_DIR)

def safe_load(path):
    # .pt 用 torch.load；.pkl 用 pickle.load
    if path.endswith('.pt'):
        return torch.load(path, weights_only=False)
    with open(path, 'rb') as f:
        return pickle.load(f)

print('safe_load ready')

# ===== 修改这里的文件名 =====
target_file = 'obs_records_20260917_175530_ep0.pkl'  # 改成你想看的文件
# =========================

path = os.path.join(BUFFER_DIR, target_file)
data = safe_load(path)

print(f'类型: {type(data)}')
print(f'长度: {len(data)}')
if data:
    print(f'第0条 keys: {list(data[0].keys())}')
    print(f'第0条 uav ids: {list(data[0]["uavs"].keys())}')

# # 需要查看的切片位置
# snap = data[177]#491 208
```
